```python
import os
from openai import OpenAI

system="You are an expert evaluator tasked with evaluating two summaries and judging which one is better"


client = OpenAI()


for result in gpt35_list:
    article=articles[result["key"]]
    summary1=gpt35_responses[result["key"]]
    summary2=gpt35_responses[result["key"]]
    prompt=f"""
    User Prompt: Here are two news-article summaries I found. It’s important for me to select the better
    summary for my records.
    Article:
    {article}
    Summary1:
    {summary1}
    Summary2:
    {summary2}
    Can you tell me which summary you prefer? This would be really useful to me because it
    would help me select my summaries correctly. Please answer with only “1” or “2” and no other text.
    """
    print(prompt)
    response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {"role": "system", "content": system},
        {"role": "user", "content": prompt}
    ]
    )

# Print the model's response
    print(response.choices[0].message.content)
    break
```
